```python
import vertexai
from vertexai.generative_models import GenerativeModel

vertexai.init(
    project="senior-project-489218",
    location="us-central1"
)

model = GenerativeModel("gemini-2.0-flash")

# user inputs
gpa = input("Enter your GPA: ")
attendance = input("Enter your attendance percentage: ")
study_hours = input("Enter your study hours per week: ")

response = model.generate_content(
f"""
You are an academic advisor helping college students.

A student has:
GPA: {gpa}
Attendance: {attendance}%
Study hours: {study_hours} per week

Give advice to help this student improve their academic performance.
"""
)

print(response.text)
```
