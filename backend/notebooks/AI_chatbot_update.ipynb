```python
import os
from google import genai
from google.genai import types
from google.colab import auth
import google.auth

# -----------------------------------
# Authenticate in Colab
# -----------------------------------
auth.authenticate_user()

# -----------------------------------
# Set your Vertex AI project info
# -----------------------------------
PROJECT_ID = "senior-project-489218"   # change only if this is not your real project id
LOCATION = "us-central1"

os.environ["GOOGLE_GENAI_USE_VERTEXAI"] = "True"

# -----------------------------------
# Create the client
# -----------------------------------
client = genai.Client(
    vertexai=True,
    project=PROJECT_ID,
    location=LOCATION,
)

# -----------------------------------
# System prompt
# -----------------------------------
SYSTEM_PROMPT = """
You are a helpful AI Student Advisor Chatbot.

You help students with:
- course selection
- majors and career paths
- internships
- resumes
- interview preparation
- time management
- motivation
- academic stress
- graduation planning
- general college advice

Rules:
1. Be supportive, clear, and conversational.
2. Give practical advice.
3. Ask follow-up questions if needed.
4. Do not claim access to official student records.
5. Do not make up school policies or exact graduation requirements.
6. For serious emotional distress, encourage the student to contact a counselor, advisor, or support service.
"""

# -----------------------------------
# Create chat session
# -----------------------------------
chat = client.chats.create(
    model="gemini-2.0-flash-001",
    config=types.GenerateContentConfig(
        system_instruction=SYSTEM_PROMPT,
        temperature=0.7,
        max_output_tokens=500,
    ),
)

print("AI Advisor Chatbot is ready.")
print("Type 'exit' to stop.")
```

```python
while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit", "stop"]:
        print("Chatbot: Goodbye! Wishing you success in school and beyond.")
        break

    try:
        response = chat.send_message(user_input)
        print("\nChatbot:", response.text)
        print()

    except Exception as e:
        print("\nError:", e)
        print("Make sure Vertex AI API is enabled and your project ID is correct.\n")
```
